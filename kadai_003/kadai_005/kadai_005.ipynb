```python
# Step1. matplotlib.pyplot と numpy をインポートする
import matplotlib.pyplot as plt
import numpy as np

# Step2. 2行3列（計6個）の描画領域（FigureとAxes）を作成する
# layout='constrained' または plt.tight_layout() を使うことでグラフ領域の重なりを防げます
fig, axes = plt.subplots(2, 3, layout="constrained")

# Step3. 各領域に指定されたグラフをプロットする

# [0, 0] 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)

# [0, 1] sin関数グラフ
x1 = np.linspace(0, 10, 200)
y1 = 2 * np.sin(2 * np.pi * x1 / 3) + 2
axes[0, 1].plot(x1, y1)
axes[0, 1].set_xlim(0, 10)
axes[0, 1].set_ylim(-1, 5)
axes[0, 1].set_xticks(range(10))

# [0, 2] ヒストグラム
np.random.seed(0)  # 再現性のためのシード固定
data_hist = np.random.normal(loc=15, scale=4, size=2000)
axes[0, 2].hist(data_hist, bins=10)

# [1, 0] 散布図
x_scatter = np.random.normal(loc=3, scale=2, size=60)
y_scatter = np.random.normal(loc=4, scale=2, size=60)
axes[1, 0].scatter(x_scatter, y_scatter)

# 残り2つの領域 [1, 1] と [1, 2] は空のままにします

# Step4. グラフを表示させる
plt.show()
```
